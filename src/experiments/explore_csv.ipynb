```python
import numpy as np
import pandas as pd
```

```python
outflow = pd.read_csv('../../data/timeseries/Outflow.txt', sep=';', parse_dates=['Date'])
radiation = pd.read_csv('../../data/timeseries/Radiation.txt', sep=';', parse_dates=['time'])
SWC = pd.read_csv('../../data/timeseries/SWC.txt', sep=';', parse_dates=['time']) 
Tsoil = pd.read_csv('../../data/timeseries/Tsoil.txt', sep=';', parse_dates=['time'])
Wetter_Campbell = pd.read_csv('../../data/timeseries/Wetter_Campbell.txt', sep=';', parse_dates=['Date'])
```

```python
for name, df in [('outflow', outflow), ('radiation', radiation), ('SWC', SWC), ('Tsoil', Tsoil), ('Wetter_Campbell', Wetter_Campbell)]:
    dups = df[df.duplicated(keep=False)].sort_values(df.columns[0])
    print(f"=== {name}: {len(dups)} rows involved in duplication ({df.duplicated().sum()} would be dropped) ===")
    display(dups.head(2))
```

```python
outflow = outflow.drop_duplicates()
radiation = radiation.drop_duplicates()
SWC = SWC.drop_duplicates()
Tsoil = Tsoil.drop_duplicates()
Wetter_Campbell = Wetter_Campbell.drop_duplicates()
len(outflow), len(radiation), len(SWC), len(Tsoil), len(Wetter_Campbell)
```
