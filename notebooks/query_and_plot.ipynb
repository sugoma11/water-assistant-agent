```python
import duckdb


with duckdb.connect("../data/water.duckdb") as con:
    one = con.execute("SELECT Sumpf1_Efflux from outflow;").fetchall()
    two = con.execute("SELECT Sumpf2_Efflux from outflow;").fetchall()
    rain = con.execute("SELECT Rain from wetter;").fetchall()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(one, label="Sumpf1_Efflux")
plt.plot(two, label="Sumpf2_Efflux")
plt.plot(rain, label="Rain", alpha=0.3)
plt.legend()
```
